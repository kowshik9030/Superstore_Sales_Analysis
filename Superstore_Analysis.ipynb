```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv("Superstore.csv", encoding="latin1")
#UTF-8 unable to read latin so by using "latin1" or "cp1252" we can over come the UniCodeDecodeError
df
```

```python
dates = ["Order Date", "Ship Date"]

df[dates] = df[dates].apply(
    lambda x: pd.to_datetime(x.str.replace("/", "-", regex=False), errors="coerce")
              .dt.strftime("%Y-%m-%d")
)
df
```

```python
df.isna().sum()
```

```python
df.describe()
```

```python
df.info()
```

```python
#type conversion....1st groupby then convert by using .astype()
DATE = ["Order Date", "Ship Date"]
df[DATE] = df[DATE].astype("datetime64[ns]")


NUM = ["Sales", "Quantity","Row ID","Postal Code"]
df[NUM] = df[NUM].astype("float64")

STR = ["Order ID", "Ship Mode", "Customer ID", "Customer Name", "Segment", "Country", "City", "Region", "Product ID", "Category", "Sub-Category", "Product Name","State"]
df[STR] = df[STR].astype("string")
```

```python
df.info()
```

```python
df["Year"] = df["Order Date"].dt.year
df
```

```python
df.info()
```

```python
#1 - total sales in each year
yearly_sales = df.groupby("Year")["Sales"].sum()
sns.lineplot(yearly_sales)    #.set(title = 'total sales in each year'
plt.title('Total sales in each year')
plt.show()
```

```python
df["Year"].unique()
```

```python
#2 - Totol profits in each year
total_profits = df.groupby("Year")["Profit"].sum()
sns.lineplot(total_profits)
plt.title("Total profits in each year")
plt.show()
```

```python
#3 - Total discounts in each year
total_discounts = df.groupby("Year")["Discount"].sum()
sns.barplot(total_discounts, width=0.5, alpha=1)
plt.title("Total discounts in each year")
#plt.grid(True)
plt.show()
```

```python
#4 - no.of customers in each segment
customer_count = df["Segment"].value_counts()
customer_count.plot(kind="bar", title="No. of Customers in Each Segment",rot=0)
plt.show()
```

```python
#5 - in each segment highest and lowest sales
segment_sales = df.groupby("Segment")["Sales"].agg(["max", "min"])
segment_sales.plot(kind="bar", title="Highest and Lowest Sales by Segment",rot = 0,logy = True)
plt.xticks(rotation=0)
plt.show()
```

```python
#6 - segment generated highest and lowest sales
#segment_sales = df.groupby("Segment")["Sales"].sum()
#segment_sales
#print("Highest segment sales are :", segment_sales.idxmax())
#print("Lowest segment sales are :", segment_sales.idxmin())
segment_sales = df.groupby("Segment")["Sales"].sum()
top_bottom = segment_sales.loc[[segment_sales.idxmax(), segment_sales.idxmin()]]
top_bottom.plot(kind="bar", title="Highest and Lowest Sales by Segment")
plt.xticks(rotation=0)
plt.show()
```

```python
#7 - total sales and profits for each year

yearly_sales = df.groupby("Year")["Sales"].sum()
yearly_profits = df.groupby("Year")["Profit"].sum()
yearly_data = [["yearly_sales","yearly_profits"]]
yearly_data = pd.DataFrame({"Sales": yearly_sales, "Profits": yearly_profits})
yearly_data.plot(kind="bar")
plt.xticks(rotation=0)
plt.show()
```

```python
#8 - relation b/w sales and profits in each segment
sns.scatterplot(data=df, x="Sales", y="Profit",alpha = 0.4,s=50)
plt.title("Sales vs Profit")
plt.show()
```

```python
#10 - region highest and lowest sales Region
region_sales = df.groupby("Region")["Sales"].sum()
top_bottom = region_sales.loc[[region_sales.idxmax(), region_sales.idxmin()]]
top_bottom.plot(kind="bar", title="Highest and Lowest Sales by Region")#,rot = 0
plt.xticks(rotation=0)
plt.show()
```

```python
#26 - total quantity ordered by Each Sub-Category in Each Region
quantity_data = df.groupby(["Region", "Sub-Category"])["Quantity"].sum().reset_index()
sns.catplot(data=quantity_data,x="Quantity",y="Sub-Category",col="Region",kind="bar",errorbar=None,height=5,aspect=0.7)
# plt.title("Total Ordered for Each Sub-Category in Each Region")

plt.xlabel("Total Quantity")
plt.ylabel("Sub-Category")
plt.suptitle("Total Ordered for Each Sub-Category in Each Region",y=1.1)
plt.show()
```

```python
#27 - for each sub-category has highest quantity sold
quantity_data = df.groupby(["Region", "Sub-Category"])["Quantity"].sum().reset_index()
highest = quantity_data.loc[quantity_data.groupby("Region")["Quantity"].idxmax()]
# sns.barplot(data=highest, x="Quantity", y="Region")
labelbar = sns.barplot(data=highest, x="Quantity", y="Region")
labelbar.bar_label(labelbar.containers[0],labels=highest["Sub-Category"],padding=1)
plt.xlabel("Highest Quantity")
plt.ylabel("Sub-Category")
plt.show()
```

```python
#28 - for each sub-category has highest quantity sold
quantity_data = df.groupby(["Region", "Sub-Category"])["Quantity"].sum().reset_index()
lowest = quantity_data.loc[quantity_data.groupby("Region")["Quantity"].idxmin()]
# sns.barplot(data=lowest, x="Quantity", y="Region")
labelbar = sns.barplot(data=lowest, x="Quantity", y="Region")
labelbar.bar_label(labelbar.containers[0],labels=lowest["Sub-Category"],padding=1)
plt.xlabel("lowest Quantity")
plt.ylabel("Sub-Category")
plt.show()
```

```python
#29 - top 10 most demanded sub-categories in each region
demand_data = df.groupby(["Region", "Sub-Category"])["Quantity"].sum().reset_index()
demand_data = (demand_data.sort_values(["Region", "Quantity"], ascending=[True, False]).groupby("Region").head(10))
sns.catplot(data=demand_data,x="Quantity",y="Sub-Category",col="Region",kind="bar",errorbar=None,height=5,aspect=0.7)
plt.xlabel("Total Quantity")
plt.ylabel("Sub-Category")
plt.show()
```

```python
#30 - visualization of top 10 most demanded sub-categories in each region
demand_data = df.groupby(["Region", "Sub-Category"])["Quantity"].sum().reset_index()
demand_data = (demand_data.sort_values(["Region", "Quantity"], ascending=[True, True]).groupby("Region").head(10))
demand_data = demand_data.sort_values("Quantity")
sns.catplot(data=demand_data,x="Quantity",y="Sub-Category",col="Region",kind="bar",errorbar=None,height=5,aspect=0.7)
plt.xlabel("Total Quantity")
plt.ylabel("Sub-Category")
plt.show()
```
